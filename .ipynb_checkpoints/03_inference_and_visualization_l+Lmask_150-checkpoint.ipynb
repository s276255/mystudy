```python
import torch
import torch.nn as nn
from transformers import CLIPTextModel, CLIPTokenizer
from diffusers import AutoencoderKL, UNet2DConditionModel, DDPMScheduler, ControlNetModel
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import os
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
import pandas as pd
from torchvision import transforms
import nibabel as nib
from tqdm.auto import tqdm
import lpips # LPIPS用
from skimage.metrics import peak_signal_noise_ratio as psnr # PSNR用
from skimage.metrics import structural_similarity as ssim # SSIM用

# --- ローカルGPUの設定 ---
os.environ['CUDA_VISIBLE_DEVICES'] = '0'
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# ==========================================
# Cell 1: ライブラリの準備
# ==========================================
import sys
import subprocess

# 足りないライブラリがあればインストール
try:
    import torchmetrics
    import lpips
except ImportError:
    print("Installing torchmetrics and lpips...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "torchmetrics", "lpips"])

import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from PIL import Image
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split
import nibabel as nib

# Diffusers
from diffusers import StableDiffusionPipeline, DPMSolverMultistepScheduler

# Metrics
from torchmetrics.image.lpip import LearnedPerceptualImagePatchSimilarity
from torchmetrics.image import PeakSignalNoiseRatio, StructuralSimilarityIndexMeasure

# Seed固定
def seed_everything(seed=42):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True

seed_everything(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# ==========================================
# Cell 2: 設定
# ==========================================
# パス設定
BASE_MODEL_ID = "runwayml/stable-diffusion-v1-5"
LORA_WEIGHTS_PATH = "/home/masaki/mystudy/trained_models_l+Lmask_150/best_model_lora" # 保存したLoRAパス
IMAGE_ROOT_DIR = '/home/masaki/mystudy/my_yale_data/yale_t1post_raw'
CSV_FILE_PATH = '/home/masaki/mystudy/my_yale_data/final_labeled_data.csv'
OUTPUT_DIR = "/home/masaki/mystudy/inference_results_lora_150" # 結果保存先

os.makedirs(OUTPUT_DIR, exist_ok=True)

IMAGE_SIZE = 256
BATCH_SIZE = 1 # 推論は1枚ずつ丁寧に
```

```python
# ==========================================
# Cell 3: データセット定義
# ==========================================
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

class BrainMetsTestDataset(Dataset):
    def __init__(self, csv_file, image_root_dir, patient_ids):
        self.data = pd.read_csv(csv_file)
        self.data = self.data[self.data['PatientID'].isin(patient_ids)]
        self.image_root_dir = image_root_dir
        self.pairs = self._create_pairs()
        
        # 評価用なので [0, 1] のテンソルにする (Metrics計算のため)
        self.transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Resize((IMAGE_SIZE, IMAGE_SIZE), antialias=True),
        ])

    def _create_pairs(self):
        pairs = []
        for _, group in self.data.groupby('PatientID'):
            baseline_scans = group[group['FinalPattern'] == 'Baseline']
            future_scans = group[group['FinalPattern'] != 'Baseline']
            if baseline_scans.empty or future_scans.empty: continue
            
            # テスト用なので、各Futureに対して1つのBaselineを紐付ける
            baseline_row = baseline_scans.iloc[0] 
            for _, future_row in future_scans.iterrows():
                pairs.append({
                    "baseline_file": baseline_row['FileName'],
                    "future_file": future_row['FileName'],
                    "condition": future_row['FinalPattern'],
                    "patient_id": future_row['PatientID']
                })
        return pairs

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        pair = self.pairs[idx]
        
        baseline_path = self._get_full_path(self.image_root_dir, pair['baseline_file'])
        future_path = self._get_full_path(self.image_root_dir, pair['future_file'])

        # 画像読み込み (0-1範囲)
        baseline_img = self._load_image(baseline_path)
        future_img = self._load_image(future_path)

        # プロンプト
        prompt = f"brain MRI, {pair['condition']}"

        return {
            "baseline_image": baseline_img, # 表示用
            "future_image": future_img,     # 正解データ (Ground Truth)
            "prompt": prompt,
            "patient_id": pair['patient_id'],
            "condition": pair['condition']
        }

    def _get_full_path(self, directory, filename):
        name = filename.replace('.nii.gz', '') + "_0000.nii.gz"
        return os.path.join(directory, name)

    def _load_image(self, path):
        nii = nib.load(path)
        data = nii.get_fdata().astype('float32')
        slice_img = data[:, :, data.shape[2] // 2]
        if slice_img.max() > slice_img.min():
            slice_img = (slice_img - slice_img.min()) / (slice_img.max() - slice_img.min())
        # ToTensorで (1, H, W) になるので (3, H, W) に複製
        return self.transform(slice_img).repeat(3, 1, 1)

print("Test Dataset defined.")
```

```python
# ==========================================
# Cell 4: テストデータの準備
# ==========================================
full_df = pd.read_csv(CSV_FILE_PATH)
patient_ids = full_df['PatientID'].unique()

# 学習時と同じ分割ロジック
train_pids, temp_pids = train_test_split(patient_ids, test_size=0.3, random_state=42)
val_pids, test_pids = train_test_split(temp_pids, test_size=0.5, random_state=42)

print(f"Test Set Patients: {len(test_pids)}")

test_dataset = BrainMetsTestDataset(CSV_FILE_PATH, IMAGE_ROOT_DIR, test_pids)
test_dataloader = DataLoader(test_dataset, batch_size=1, shuffle=False)

print(f"Total Test Images: {len(test_dataset)}")
```

```python
# ==========================================
# Cell 5: モデルロード & LoRA適用
# ==========================================
print("Loading Stable Diffusion Pipeline...")
pipe = StableDiffusionPipeline.from_pretrained(
    BASE_MODEL_ID, 
    torch_dtype=torch.float32,
    safety_checker=None # 医療画像で誤作動しないよう無効化
)

# LoRAのロード
print(f"Loading LoRA weights from: {LORA_WEIGHTS_PATH}")
pipe.load_lora_weights(LORA_WEIGHTS_PATH)

# スケジューラの設定 (推論を高速・高品質に)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe.to(device)

print("✅ Model loaded successfully (LoRA w/o ControlNet)")
```

```python
# ==========================================
# Cell 6: 評価指標の準備
# ==========================================
# LPIPS (VGG使用, 値が低いほど良い)
calc_lpips = LearnedPerceptualImagePatchSimilarity(net_type='vgg').to(device)

# SSIM (値が高いほど良い, 1.0が最大)
calc_ssim = StructuralSimilarityIndexMeasure(data_range=1.0).to(device)

# PSNR (値が高いほど良い)
calc_psnr = PeakSignalNoiseRatio(data_range=1.0).to(device)

print("Metrics initialized.")
```

```python
# ==========================================
# Cell 7: 推論 & 定量評価ループ
# ==========================================
results = []
generated_images = [] # 後で可視化するためにいくつか保存

print("Starting Inference...")

for i, batch in enumerate(tqdm(test_dataloader)):
    # データの準備
    baseline = batch["baseline_image"].to(device) # (B, 3, H, W) [0,1]
    ground_truth = batch["future_image"].to(device) # (B, 3, H, W) [0,1]
    prompt = batch["prompt"][0]
    pid = batch["patient_id"][0]
    cond = batch["condition"][0]

    # --- 1. 画像生成 ---
    # ControlNetがないので、テキストのみで生成
    # seedを固定して再現性を確保
    generator = torch.Generator(device).manual_seed(42 + i)
    
    with torch.no_grad():
        # inference_steps=30〜50くらいが標準
        image = pipe(prompt, num_inference_steps=30, generator=generator).images[0]

    # --- 2. 前処理 (生成画像をTensor変換) ---
    image = image.resize((IMAGE_SIZE, IMAGE_SIZE))
    pred_tensor = transforms.ToTensor()(image).unsqueeze(0).to(device) # (1, 3, H, W), [0,1]

    # --- 3. スコア計算 ---
    # LPIPSは入力が [-1, 1] であることを期待する場合があるが、torchmetricsは [0,1] で正規化設定可能
    # ここではデフォルト設定に合わせて [0,1] -> [-1,1] にして渡すのが安全
    gt_norm = ground_truth * 2.0 - 1.0
    pred_norm = pred_tensor * 2.0 - 1.0
    
    lpips_score = calc_lpips(pred_norm, gt_norm).item()
    ssim_score = calc_ssim(pred_tensor, ground_truth).item()
    psnr_score = calc_psnr(pred_tensor, ground_truth).item()

    # 結果保存
    results.append({
        "PatientID": pid,
        "Condition": cond,
        "SSIM": ssim_score,
        "PSNR": psnr_score,
        "LPIPS": lpips_score
    })

    # 可視化用に最初の5枚だけ保存
    if i < 5:
        generated_images.append({
            "baseline": baseline.cpu(),
            "ground_truth": ground_truth.cpu(),
            "prediction": pred_tensor.cpu(),
            "condition": cond,
            "patient_id": pid
        })

# --- 平均スコアの計算 ---
df_results = pd.DataFrame(results)
print("\n=== Quantitative Results (LoRA w/o ControlNet) ===")
print(df_results[["SSIM", "PSNR", "LPIPS"]].mean())

# CSV保存
df_results.to_csv(os.path.join(OUTPUT_DIR, "quantitative_results.csv"), index=False)
print(f"Results saved to {OUTPUT_DIR}")
```

```python
# ==========================================
# Cell 8: 定性評価 (Visualization)
# ==========================================
def show_comparison(samples):
    n = len(samples)
    fig, axes = plt.subplots(n, 3, figsize=(12, 4 * n))
    
    if n == 1: axes = [axes] # 1枚の時の対応

    for i, sample in enumerate(samples):
        # Tensor -> Numpy (H, W, C)
        base = sample["baseline"][0].permute(1, 2, 0).numpy()
        gt = sample["ground_truth"][0].permute(1, 2, 0).numpy()
        pred = sample["prediction"][0].permute(1, 2, 0).numpy()

        # 表示 (Baseline)
        ax0 = axes[i][0] if n > 1 else axes[0]
        ax0.imshow(base, cmap="gray")
        ax0.set_title(f"Baseline MRI\n({sample['patient_id']})")
        ax0.axis("off")

        # 表示 (Generated - LoRA)
        ax1 = axes[i][1] if n > 1 else axes[1]
        ax1.imshow(pred, cmap="gray")
        ax1.set_title(f"Generated (LoRA Only)\n{sample['condition']}")
        ax1.axis("off")

        # 表示 (Ground Truth)
        ax2 = axes[i][2] if n > 1 else axes[2]
        ax2.imshow(gt, cmap="gray")
        ax2.set_title(f"Ground Truth\n{sample['condition']}")
        ax2.axis("off")

    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "qualitative_comparison.png"))
    plt.show()

# 保存しておいた画像を表示
show_comparison(generated_images)
```

```python
# ==========================================
# Cell 7: 推論 & 定量評価ループ (条件別集計版)
# ==========================================
results = []
generated_images = [] # 後で可視化するためにいくつか保存

print("Starting Inference...")

for i, batch in enumerate(tqdm(test_dataloader)):
    # データの準備
    baseline = batch["baseline_image"].to(device) # (B, 3, H, W) [0,1]
    ground_truth = batch["future_image"].to(device) # (B, 3, H, W) [0,1]
    prompt = batch["prompt"][0]
    pid = batch["patient_id"][0]
    cond = batch["condition"][0]

    # --- 1. 画像生成 ---
    # ControlNetがないので、テキストのみで生成
    # seedを固定して再現性を確保
    generator = torch.Generator(device).manual_seed(42 + i)
    
    with torch.no_grad():
        # inference_steps=30〜50くらいが標準
        image = pipe(prompt, num_inference_steps=30, generator=generator).images[0]

    # --- 2. 前処理 (生成画像をTensor変換) ---
    image = image.resize((IMAGE_SIZE, IMAGE_SIZE))
    pred_tensor = transforms.ToTensor()(image).unsqueeze(0).to(device) # (1, 3, H, W), [0,1]

    # --- 3. スコア計算 ---
    # LPIPSは入力が [-1, 1] であることを期待する場合があるが、torchmetricsは [0,1] で正規化設定可能
    # ここではデフォルト設定に合わせて [0,1] -> [-1,1] にして渡すのが安全
    gt_norm = ground_truth * 2.0 - 1.0
    pred_norm = pred_tensor * 2.0 - 1.0
    
    lpips_score = calc_lpips(pred_norm, gt_norm).item()
    ssim_score = calc_ssim(pred_tensor, ground_truth).item()
    psnr_score = calc_psnr(pred_tensor, ground_truth).item()

    # 結果保存
    results.append({
        "PatientID": pid,
        "Condition": cond,
        "SSIM": ssim_score,
        "PSNR": psnr_score,
        "LPIPS": lpips_score
    })

    # 可視化用に最初の5枚だけ保存
    if i < 5:
        generated_images.append({
            "baseline": baseline.cpu(),
            "ground_truth": ground_truth.cpu(),
            "prediction": pred_tensor.cpu(),
            "condition": cond,
            "patient_id": pid
        })

# --- 集計と表示 (ここを変更しました) ---
df_results = pd.DataFrame(results)

# 1. 全体の平均 (Overall)
print("\n=== Overall Quantitative Results (LoRA w/o ControlNet) ===")
print(df_results[["SSIM", "PSNR", "LPIPS"]].mean())

# 2. 条件ごとの平均 (By Condition)
print("\n=== Results by Condition (Progression vs Favorable) ===")
# Condition列でグループ化して平均を算出
grouped_results = df_results.groupby("Condition")[["SSIM", "PSNR", "LPIPS"]].mean()
print(grouped_results)

# --- CSV保存 ---
# 全データの保存
df_results.to_csv(os.path.join(OUTPUT_DIR, "quantitative_results_all.csv"), index=False)

# 条件別平均の保存
grouped_results.to_csv(os.path.join(OUTPUT_DIR, "quantitative_results_by_condition.csv"))

print(f"Results saved to {OUTPUT_DIR}")
```
