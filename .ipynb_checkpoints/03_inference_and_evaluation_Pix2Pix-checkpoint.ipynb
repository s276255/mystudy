```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader # ← Dataset を追加！
from torchvision import transforms
from tqdm.auto import tqdm
import os
import pandas as pd
import numpy as np
import nibabel as nib
from PIL import Image
import matplotlib.pyplot as plt

# 評価用ライブラリ
import lpips
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
from torchmetrics.image.fid import FrechetInceptionDistance

# --- デバイスの設定 ---
os.environ['CUDA_VISIBLE_DEVICES'] = '0'
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# --- パスの設定 ---
IMAGE_DIR = '/home/masaki/mystudy/my_yale_data/yale_t1post_raw'
CSV_FILE = '/home/masaki/mystudy/my_yale_data/final_labeled_data.csv'
MODEL_PATH = "/home/masaki/mystudy/trained_models_masked_loss_+Pix2Pix/trained_models_masked_loss_+Pix2Pix.pth"

# 生成画像の保存先（この後ここを読み取って評価します）
SAVE_DIR = "./generated_images_pix2pix"
os.makedirs(SAVE_DIR, exist_ok=True)

IMAGE_SIZE = 256
BATCH_SIZE = 1 # 推論時は1枚ずつ処理するのが確実です
```

```python
from sklearn.model_selection import train_test_split

full_df = pd.read_csv(CSV_FILE)
patient_ids = full_df['PatientID'].unique()

# 学習時と同じrandom_state=42を使用
train_patient_ids, temp_patient_ids = train_test_split(
    patient_ids, test_size=0.3, random_state=42
)
val_patient_ids, test_patient_ids = train_test_split(
    temp_patient_ids, test_size=0.5, random_state=42
)

print(f"Test patients: {len(test_patient_ids)} (Used for inference)")
```

```python
class BrainMetsDataset(Dataset):
    def __init__(self, csv_file, image_root_dir, patient_ids, mode, image_size=256):
        self.data = pd.read_csv(csv_file)
        self.data = self.data[self.data['PatientID'].isin(patient_ids)]
        self.image_root_dir = image_root_dir
        self.mode = mode
        self.image_size = image_size
        self.pairs = self._create_pairs()
        
        self.transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Resize((self.image_size, self.image_size), antialias=True),
            transforms.Normalize([0.5], [0.5])
        ])

    def _create_pairs(self):
        pairs = []
        for _, group in self.data.groupby('PatientID'):
            baseline_scans = group[group['FinalPattern'] == 'Baseline']
            future_scans = group[group['FinalPattern'] != 'Baseline']
            if baseline_scans.empty or future_scans.empty: continue
            for _, baseline_row in baseline_scans.iterrows():
                for _, future_row in future_scans.iterrows():
                    pairs.append({
                        "baseline_file": baseline_row['FileName'],
                        "future_file": future_row['FileName'],
                        "condition": future_row['FinalPattern']
                    })
        print(f"Found {len(pairs)} pairs for test.")
        return pairs

    def __len__(self): return len(self.pairs)

    def __getitem__(self, idx):
        pair = self.pairs[idx]
        baseline_path = self._get_full_path(self.image_root_dir, pair['baseline_file'])
        future_path = self._get_full_path(self.image_root_dir, pair['future_file'])

        baseline_image = self._load_and_process_image(baseline_path)
        future_image = self._load_and_process_image(future_path)

        return {
            "baseline_image": baseline_image,
            "future_image": future_image,
            "condition_label": pair['condition'],
            "baseline_filename": pair['baseline_file'].replace('.nii.gz', '')
        }
            
    def _get_full_path(self, directory, filename):
        name_without_ext = filename.replace('.nii.gz', '')
        return os.path.join(directory, f"{name_without_ext}_0000.nii.gz")

    def _load_and_process_image(self, path):
        nii = nib.load(path)
        data = nii.get_fdata().astype('float32')
        slice_img = data[:, :, data.shape[2] // 2]
        if slice_img.max() > slice_img.min():
            slice_img = (slice_img - slice_img.min()) / (slice_img.max() - slice_img.min())
        return self.transform(slice_img).repeat(3, 1, 1)

test_dataset = BrainMetsDataset(CSV_FILE, IMAGE_DIR, test_patient_ids, mode='test')
test_dataloader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)
```

```python
# 学習時と同じGenerator構造を定義
class UNetDown(nn.Module):
    def __init__(self, in_size, out_size, normalize=True, dropout=0.0):
        super().__init__()
        layers = [nn.Conv2d(in_size, out_size, 4, 2, 1, bias=False)]
        if normalize: layers.append(nn.BatchNorm2d(out_size))
        layers.append(nn.LeakyReLU(0.2))
        if dropout: layers.append(nn.Dropout(dropout))
        self.model = nn.Sequential(*layers)
    def forward(self, x): return self.model(x)

class UNetUp(nn.Module):
    def __init__(self, in_size, out_size, dropout=0.0):
        super().__init__()
        layers = [nn.ConvTranspose2d(in_size, out_size, 4, 2, 1, bias=False),
                  nn.BatchNorm2d(out_size), nn.ReLU(inplace=True)]
        if dropout: layers.append(nn.Dropout(dropout))
        self.model = nn.Sequential(*layers)
    def forward(self, x, skip_input):
        x = self.model(x)
        return torch.cat((x, skip_input), 1)

class GeneratorUNet(nn.Module):
    def __init__(self, in_channels=4, out_channels=3):
        super().__init__()
        self.down1 = UNetDown(in_channels, 64, normalize=False)
        self.down2 = UNetDown(64, 128)
        self.down3 = UNetDown(128, 256)
        self.down4 = UNetDown(256, 512, dropout=0.5)
        self.down5 = UNetDown(512, 512, dropout=0.5)
        self.down6 = UNetDown(512, 512, dropout=0.5)
        self.down7 = UNetDown(512, 512, dropout=0.5)
        self.down8 = UNetDown(512, 512, normalize=False, dropout=0.5)
        self.up1 = UNetUp(512, 512, dropout=0.5)
        self.up2 = UNetUp(1024, 512, dropout=0.5)
        self.up3 = UNetUp(1024, 512, dropout=0.5)
        self.up4 = UNetUp(1024, 512, dropout=0.5)
        self.up5 = UNetUp(1024, 256)
        self.up6 = UNetUp(512, 128)
        self.up7 = UNetUp(256, 64)
        self.final = nn.Sequential(nn.ConvTranspose2d(128, out_channels, 4, 2, 1), nn.Tanh())
    def forward(self, x):
        d1 = self.down1(x); d2 = self.down2(d1); d3 = self.down3(d2); d4 = self.down4(d3)
        d5 = self.down5(d4); d6 = self.down6(d5); d7 = self.down7(d6); d8 = self.down8(d7)
        u1 = self.up1(d8, d7); u2 = self.up2(u1, d6); u3 = self.up3(u2, d5); u4 = self.up4(u3, d4)
        u5 = self.up5(u4, d3); u6 = self.up6(u5, d2); u7 = self.up7(u6, d1)
        return self.final(u7)

# モデルロード
generator = GeneratorUNet().to(device)
generator.load_state_dict(torch.load(MODEL_PATH))
generator.eval()
print("Pix2Pix Generator loaded successfully.")
```

```python
print(f"Starting Inference. Images will be saved to {SAVE_DIR}")

with torch.no_grad():
    for batch in tqdm(test_dataloader, desc="Inference"):
        real_base = batch["baseline_image"].to(device)
        condition = batch["condition_label"][0]
        filename = batch["baseline_filename"][0]
        
        # 入力作成
        b, _, h, w = real_base.shape
        val = 1.0 if condition == "Progression" else -1.0
        label_img = torch.full((1, 1, h, w), val, device=device)
        input_gen = torch.cat([real_base, label_img], 1)
        
        # 生成
        fake_future = generator(input_gen)
        
        # [-1, 1] -> [0, 255] uint8 に変換して保存
        gen_np = (fake_future[0, 0].cpu().numpy() / 2 + 0.5).clip(0, 1) * 255.0
        img = Image.fromarray(gen_np.astype(np.uint8))
        
        # 保存名：条件_ファイル名.png (後で評価時に条件を識別するため)
        save_name = f"{condition}_{filename}.png"
        img.save(os.path.join(SAVE_DIR, save_name))

print("🎉 All test images saved successfully!")
```

```python
print("\n--- Starting Quantitative Evaluation from Saved Images ---")

# メトリクスの初期化
loss_fn_lpips = lpips.LPIPS(net='alex').to(device)
fid_prog = FrechetInceptionDistance(feature=2048).to(device)
fid_fav = FrechetInceptionDistance(feature=2048).to(device)

def to_fid_img(tensor):
    img = (tensor / 2 + 0.5).clamp(0, 1) * 255.0
    return img.to(torch.uint8).repeat(1, 3, 1, 1) if img.shape[1] == 1 else img.to(torch.uint8)

prog_metrics = {'lpips': [], 'psnr': [], 'ssim': []}
fav_metrics = {'lpips': [], 'psnr': [], 'ssim': []}

# データセットを再ループして、保存した画像とGTを比較
for batch in tqdm(test_dataloader, desc="Calculating Metrics"):
    real_future = batch["future_image"].to(device)
    condition = batch["condition_label"][0]
    filename = batch["baseline_filename"][0]
    
    # 保存した画像を読み込む
    gen_path = os.path.join(SAVE_DIR, f"{condition}_{filename}.png")
    gen_pil = Image.open(gen_path).convert('RGB')
    gen_tensor = transforms.ToTensor()(gen_pil).unsqueeze(0).to(device) * 2 - 1 # [0,1] -> [-1,1]

    # LPIPS
    l_score = loss_fn_lpips(real_future, gen_tensor).item()
    
    # PSNR/SSIM (0-1スケール)
    real_np = (real_future[0, 0].cpu().numpy() / 2 + 0.5).clip(0, 1)
    gen_np = (gen_tensor[0, 0].cpu().numpy() / 2 + 0.5).clip(0, 1)
    p_score = psnr(real_np, gen_np, data_range=1.0)
    s_score = ssim(real_np, gen_np, data_range=1.0)
    
    # FID蓄積
    real_fid = to_fid_img(real_future)
    gen_fid = to_fid_img(gen_tensor)

    if condition == "Progression":
        prog_metrics['lpips'].append(l_score); prog_metrics['psnr'].append(p_score); prog_metrics['ssim'].append(s_score)
        fid_prog.update(real_fid, real=True); fid_prog.update(gen_fid, real=False)
    else:
        fav_metrics['lpips'].append(l_score); fav_metrics['psnr'].append(p_score); fav_metrics['ssim'].append(s_score)
        fid_fav.update(real_fid, real=True); fid_fav.update(gen_fid, real=False)

# 結果表示
def print_res(name, m, fid_obj):
    print(f"\n--- {name} Category ---")
    print(f"LPIPS (↓): {np.mean(m['lpips']):.4f} ± {np.std(m['lpips']):.4f}")
    print(f"PSNR  (↑): {np.mean(m['psnr']):.4f} ± {np.std(m['psnr']):.4f}")
    print(f"SSIM  (↑): {np.mean(m['ssim']):.4f} ± {np.std(m['ssim']):.4f}")
    print(f"FID   (↓): {fid_obj.compute().item():.4f}")

print_res("Progression", prog_metrics, fid_prog)
print_res("Favorable Outcome", fav_metrics, fid_fav)
```
